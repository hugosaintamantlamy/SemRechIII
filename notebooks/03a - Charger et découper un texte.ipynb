# Analyse de texte littéraire : Charger et découper un texte

Dans ce notebook, nous verrons comment charger un fichier (ou plusieurs fichiers), puis comment commencer à le/les découper (tokenisation).



## 1. Chargement d'un texte

Jusqu'ici, nous avons toujours travaillé sur des chaînes inscrites directement dans le code. C'est une option adéquate pour de courts extraits de textes, mais ce n'est évidemment pas viable pour explorer des ensembles de données plus larges (ouvrage complet, corpus). 

La solution à ce problème est de charger dans Python les fichiers qui nous intéressent, et que nous avons préalablement enregistrés sur notre machine. 

Charger un fichier texte revient à mettre en mémoire son contenu afin de l'utiliser dans Python.

### 1.1 Chargement d'un seul fichier

Il existe une panoplie de formats de fichiers qui peuvent contenir du texte (DOCX, PDF, EPUB, TXT...).

Le format TXT (Plain Text) est le plus simple et le plus universel. Il n'inclut aucune mise en forme (pas de polices différentes, pas de gras/italique, pas de couleurs, etc.), ce qui facilite son traitement. Il est préférable de travailler avec des fichiers qui ont ce format. Au besoin, il est possible de convertir d'autres types de fichiers au format TXT.

Des éditeurs de texte permettent d'ouvrir et de modifier facilement des fichiers TXT. Quelques choix populaires :
- **TextEdit** : éditeur de texte de base pour Mac
- **Bloc-notes** : éditeur de texte de base pour Windows
- **BBEdit** : éditeur de texte plus avancé pour Mac (https://www.barebones.com/products/bbedit/)
- **Notepad++** : éditeur de texte plus avancé pour Windows (https://notepad-plus-plus.org/)
- Les éditeurs plus avancés permettent notamment d'afficher les numéros de lignes. 

#### Charger son fichier dans l'environnement Jupyter
- Pour que Python ait accès au fichier, il faut que celui-ci se trouve dans l'environnement Jupyter que l'on utilise
- Si votre environnement Jupyter est sur votre machine (ex. via Anaconda), vous avez accès à tous vos fichiers
- Si vous utilisez un environnement virtuel (ex. Binder, jupyter.org/try), il faut d'abord charger son fichier dans cet environnement
- Pour y arriver, faire **Upload files** (petite flèche vers le haut dans la barre latérale) et aller chercher le fichier
- Le fichier texte apparaît alors dans la liste des fichiers

#### Charger son fichier dans Python
- On peut alors importer le fichier dans Python via la commande `open`


```python
# Exemple de code pour charger un fichier
# (Attention : Ce code ne s'exécutera que si le fichier existe)

# ouvrir le fichier
with open('Flaubert_1857 - Madame Bovary.txt', 'r', encoding='utf-8') as fichier:
    # assigner un nom de variable au contenu textuel
    texte = fichier.read()

# trouver et imprimer le nombre de caractères dans le fichier chargé
nb_caracteres = len(texte)
print("Nombre de caractères dans le fichier :", nb_caracteres)
```

#### Explication du code : 
- open('Flaubert_1857 - Madame Bovary.txt', 'r', encoding='utf-8') : ouvre le fichier en mode **lecture** ('r' pour *read*)
- encoding='utf-8' : spécifie l'encodage (essentiel pour les accents du français)
- with ... as fichier: : ouvre le fichier et le ferme automatiquement après usage
- fichier.read() : lit tout le contenu du fichier en une seule chaîne

**Emplacement du fichier** :
- Si votre fichier est dans le même dossier que votre notebook, vous pouvez l'ouvrir en utilisant simplement son titre et son extension (.txt)
- Si vous utilisez une interface Jupyter local (ex. Acaconda), vous pouvez ouvrir un fichier ouvert n'importe où sur votre machine en spécifiant l'emplacement (voir plus bas)


```python
# ouvrir le fichier dans un emplacement particulier de la machine
# ajouter petit 'r' devant la chaîne qui spécifie l'emplacement + titre + extension
# le 'r' permet d'éviter certains problèmes associés au caractère "\" (Windows seulement)

emplacement = r""

with open(emplacement, 'r', encoding='utf-8') as fichier:
    # assigner un nom de variable au contenu textuel
    texte = fichier.read()

# trouver et imprimer le nombre de caractères dans le fichier chargé
nb_caracteres = len(texte)
print("Nombre de caractères dans le fichier :", nb_caracteres)
```

## 1.2 Chargement de plusieurs fichiers

Pour effectuer des analyses comparatives, il est utile de pouvoir charger plusieurs fichiers et de les traiter ensemble. Par exemple, on peut préparer un code que l'on applique à tous les fichiers d'une liste de fichiers avec une boucle `for`.

**Étapes pour charger un groupe de fichiers**
- Constituer une liste de chemins de fichiers (chemin = emplacement + nom du fichier + extension)
- Charger les fichiers

```python
# CHARGER PLUS D'UN FICHIER - ÉTAPE 1 : CONSTITUER LISTE DE FICHIERS

# importer la bibliothèque os, qui facilite le traitement des fichiers
import os

# Fonction qui crée une liste de fichier
def lister_fichiers(dossier, contient):

    # Créer une liste vide où mettre les chemins
    fichiers = []

    # Passer à travers le dossier indiqué 
    for nom_fichier in os.listdir(dossier):
        
        # S'assurer que le nom de fichier contient la chaîne spécifiée (ici, ".txt") 
        if contient in nom_fichier:
            # Ajouter le chemin complet
            chemin_complet = os.path.join(dossier, nom_fichier)
            fichiers.append(chemin_complet)
    
    # Trier la liste en ordre alphabétique
    fichiers.sort()

    # Retourner la liste de fichiers
    return(fichiers)



# Spécifier le dossier où aller chercher les fichiers (utiliser 'r' devant le chemin) +
# critère à remplir (ici, doit contenir la chaîne ".txt")
liste_fichiers = lister_fichiers(r"C:\Users\sainhu01\OneDrive - UQAR\Documents\Cours\2025 - Données textuelles\Environnement\Données", ".txt")

# Si les fichiers que vous voulez sont dans le même dossier que votre notebook, utiliser le chemin ".
#liste_fichiers = lister_fichiers("Corpus", ".txt")

# Si vous utilisez un environnement virtuel (Binder, jupyter.org/try), vous pouvez créer des sous-dossiers (New folder) et y référer

# Passer à travers la liste de fichiers et imprimer l'emplacement de chacun
for fichier in liste_fichiers:
    print(fichier)

# Possible de retirer ou ajouter des éléments manuellement. P. ex. :
#liste_fichier.remove(r"C:\Users\sainhu01\OneDrive - UQAR\Documents\Cours\2025 - Données textuelles\Environnement\Données\notes.txt")
```

```python
# CHARGER PLUS D'UN FICHIER - ÉTAPE 2 : CHARGER LES FICHIERS

# Une fois que nous avons notre liste de fichiers, 
# nous pouvons les ouvrir avec la méthode vue plus haut


# Passer à travers la liste des fichiers
for fichier in liste_fichiers :
    # extraire le nom du fichier (sans l'emplacement)
    titre = os.path.basename(fichier)
    print(titre)

    # ouvrir le fichier (même méthode que plus haut)
    with open(fichier, 'r', encoding='utf-8') as f:
        # assigner un nom de variable au contenu textuel
        texte = f.read()
    
    # trouver et imprimer le nombre de caractères dans chaque fichier
    nb_caracteres = len(texte)
    print("- Nombre de caractères dans le texte :", nb_caracteres, "\n")
```

## 2. La tokenisation

Maintenant que nous savons charger des textes, nous pouvons passer à l'extraction des mots qu'ils contiennent.
Cette opération s'appelle la **tokenisation** (de l'anglais *token* = unité).

Un **token** est une unité textuelle, généralement un mot, mais parfois aussi :
- Un signe de ponctuation
- Un nombre
- Un symbole

L'objectif de la tokenisation est de séparer le texte en unités distinctes, que nous pourrons ensuite recenser, calculer, analyser, etc.

### 2.1 La fonction `split()`

Jusqu'ici, nous avons utilisé une méthode rudimentaire pour séparer une chaîne en mots, soit `split()`, qui extrait tous les éléments séparés par des espaces.

Cette approche a toutefois certaines limites. Regardons la liste de mots qui est produite par le code suivant : 

```python
# Premières lignes du roman Salammbô, de Gustave Flaubert
salammbo = "C'était à Mégara, faubourg de Carthage, dans les jardins d'Hamilcar. Les soldats qu'il avait commandés en Sicile se donnaient un grand festin pour célébrer le jour anniversaire de la bataille d'Éryx, et comme le maître était absent et qu'ils se trouvaient nombreux, ils mangeaient et ils buvaient en pleine liberté."

for mot in salammbo.split():
    print(mot)
```

## 2.2 Le problème du « mot »

La définition du mot comme un élément indépendant isolé des autres par des espaces est malheureusement trop simpliste. 

En plus des problèmes relevés plus haut, on peut considérer ceux qui suivent : 
- **Statut du trait d'union** : Les éléments séparés par un trait d'union forment-il un ou plusieurs mots (p. ex. *grand-mère, arc-en-ciel, dit-il, pensai-je*)
- **Mots composés** : Certains mots composés sont séparés par une espace (p. ex. *noeud papillon, bande dessinée, Moyen Âge, Thérèse Raquin*). S'agit-il d'un ou de plusieurs mots?
- **Point dans des mots** : Il est courant d'analyser un texte phrase par phrase, souvent en utilisant le point comme séparateur. Or, que faire avec des mots qui comprennent un point (p. ex. *M. Zola, S.A.Q., William S. Burrough*)?
