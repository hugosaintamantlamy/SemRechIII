## 1. LA TOKENISATION

### Qu'est-ce que la tokenisation?

La **tokenisation** est l'opération qui consiste à découper un texte en unités linguistiques distinctes, appelées **tokens**. Un token peut être :
- Un mot
- Un signe de ponctuation
- Un nombre
- Un symbole

### Pourquoi tokeniser?

La tokenisation est la première étape de toute analyse textuelle. Elle détermine :
- Ce que nous comptons comme « mot »
- Comment nous gérons les cas complexes (apostrophes, traits d'union, etc.)
- La qualité de toutes les analyses ultérieures

### Une opération complexe...

Comme nous avons vu au dernier cours, la tokenisation soulève de nombreuses questions. Par exemple : 
- Que faire avec « l'homme »? Un seul token ou deux (« l' » + « homme »)?
- « c'est-à-dire » est-il un mot ou plusieurs ?
- « M. Dupont » : comment séparer le point de l'abréviation du point de fin de phrase ?
- Faut-il traiter « Paris » et « paris » comme le même mot?

La tokenisation « naïve », avec l'espace comme séparateur, donne des résultats insatisfaisants. Nous allons nous intéresser aujourd'hui aux deux principaux problèmes : les **apostrophes** et les **traits d'union**.

### Différentes approches pour différentes questions

- Il n'existe pas une façon universelle de tokeniser un texte.
- Dépend des questions de recherche.
- Par exemple, si vous ne vous intéressez pas explicitement aux noms propres, la distinction entre majuscules et minuscules n'est peut-être pas essentielle
- Si vous vous intéressez seulement au contenu lexical d'un texte, la ponctuation peut peut-être être retirée complètement
- etc.

### 1.2. La ponctuation

Le français comprend 9 principaux signes de ponctuation :

- point [ . ]
- virgule [ , ]
- point d'exclamation [ ! ]
- point d'interrogation [ ? ]
- point de suspension [ ... ] / [ … ]
- deux-points [ : ]
- point-virgule [ ; ]
- guillemets [ « » ]
- parenthèses [ ( ) ]

Nous voulons nous assurer de distinguer les mots et les symboles de ponctuation lors de la segmentation. Deux approches simples peuvent être adoptées :
1. Si nous voulons exclure la ponctuation des tokens : effacer ces symboles
2. Si nous voulons inclure les signes de ponctuation comme tokens à part entière : ajouter espaces

```python
# Traitement de la ponctuation pour la tokenisation


# MÉTHODE 1 : effacer ponctuation

# remplacer ponctuation par rien
nana_metho1 = nana.replace(".", "").replace(",", "").replace("!", "").replace("?", "").replace("...", "").replace("…", "").replace(":", "").replace(";", "").replace("«", "").replace("»", "").replace("(", "").replace(")", "")
print("###### Méthode 1 : effacer ponctuation ######\n\n", nana_metho1)




# MÉTHODE 2 : ajouter espaces

# remplacer ponctuation par ponctuation + espace
nana_metho2 = nana.replace(".", " .").replace(",", " ,").replace("!", " !").replace("?", " ?").replace("...", " ...").replace("…", " …").replace(":", " :").replace(";", " ;").replace("«", "« ").replace("»", " »").replace("(", "( ").replace(")", " )")

# effacer espaces doubles qui auraient pu être créées par procédure précédente
nana_metho2 = nana_metho2.replace("  ", " ")

print("\n\n\n###### Méthode 2 : ajouter espace avant/après ponctuation ######\n\n", nana_metho2)


```

### 1.3. Les apostrophes

Il existe deux sortes d'apostrophes en français, le droit `'` et le courbe `’`. On trouve de la variation selon les textes. En cas de doute, il vaut mieux inclure les deux dans la procédure.

#### Deux usages de l'apostrophe
1. Pour séparer un clitique réduit du mot qui suit (p. ex. *j'arrive, qu'importe, n'entends, l'aurore*...)
2. Comme élément à part entière d'un mot indépendant (p. ex. *aujourd'hui, entr'acte, presqu'île*...)

#### Comment tokeniser?
- Il serait étrange de traiter *le chien* comme deux mots, mais *l'oiseau* comme un seul. Il est donc assez important d'inclure une procédure pour traiter le premier type d'apostrophes.
- Les apostrophes de la deuxième catégorie sont assez rares, mais une procédure trop puissante, qui traiterait *aujourd* et *hui* comme deux mots, n'est pas souhaitable.
- Une solution possible serait de placer une espace après une apostrophe si (et seulement si) le mot commence par ce qui semble être un clitique réduit (*c d j l m n qu s t*)



```python
# TRAITEMENT DIFFÉRENCIÉ DES APOSTROPHES

# créer une liste où placer les éléments (modifiés ou non)
nana_liste = []

# liste de clitiques réduits possibles
clitiques = ["c", "d", "j", "l", "m", "n", "qu", "s", "t", "C", "D", "J", "L", "M", "N", "Qu", "S", "T"]

# uniformiser les apostrophes dans la chaîne qui nous intéresse 
nana_unif = nana_metho2.replace("'", "’")

# passer à travers chaque mot de la chaîne (segmentation préliminaire avec espace)
for mot in nana_unif.split():

    # si apostrophe dans mot, vérifier ce qui vient avant l'apostrophe (avec split)
    if "’" in mot :
        avant_apos = mot.split("’")[0]
        # si ce qui vient avant l'apostrophe est dans la liste de clitiques, ajouter espace après apostrophe
        if avant_apos in clitiques :
            mot = mot.replace("’", "’ ")

    # ajouter le mot (modifié ou non) à la liste finale d'éléments
    nana_liste.append(mot)
    
# tranformer nana_liste en chaîne avec join()
nana2 = " ".join(nana_liste)
print(nana2)
```

### 1.4. Les tirets 

Il existe trois sortes de tirets en français : 

- **tiret court (quart de cadratin)** : [ - ] (trait d'union)
- **tiret moyen (demi cadratin)** : [ – ] (répliques de théâtre, propositions incises, intervalles, compositions complexes : p. ex. *Saguenay–Lac-Saint-Jean*)
- **tirer long (cadratin)** : [ — ] (répliques de théâtre, propositions incises)

Les tirets longs et moyens s'utilisent très rarement à l'intérieur des mots, nous pourrions donc les traiter comme les signes de ponctuation vus plus haut :
- ajouter `.replace("—", "").replace("–", "")` à la méthode 1 (effacer ponctuation)
- ajouter `.replace("—", " — ").replace("–", " – ")` à la méthode 2 (ajouter espace) 

#### Que faire avec le trait d'union
Le trait d'union est plus problématique. Ses usages sont multiples : 
1. Devant pronoms personnels postposés, avec ou sans « t » euphonique (p. ex. *espérait-il, demanda-t-il, sais-tu, prendrai-je*...).
2. Devant démonstatifs (p. ex. *marionettes-là, pays-ci*...)
3. Devant adjectifs emphatiques (p. ex. *lui-même, vous-mêmes...*)
4. Des cas clairs de mots composés (p. ex. *café-concert, eau-de-vie, tam-tam, sang-froid*...)
5. Des cas plus ambigus (p. ex. *par-dessus, quelques-uns*...)

On pourrait débattre des cas qui sont à traiter comme une suite de deux mots et des cas qui sont à traiter comme des mots composés indépendants. Ici, nous allons nous concentrer sur les points (1) et (2). 

#### Méthode
- Nous pouvons procéder comme avec les apostrophes : traiter chaque mot séparément, puis effacer l'apostrophe si ce qui le suit est un pronom ou un démonstratif.
- Pour se débarrasser des « t » euphoniques, on peut remplacer les séquences « -t- » et « -t' » par ` t_`, ou même par rien. 

```python
# créer une liste où placer les éléments (modifiés ou non)
nana_liste2 = []

# liste de clitiques réduits possibles
pron_dem = ["je", "tu", "il", "elle", "nous", "vous", "ils", "elles", "on", "moi", "toi", "lui", "leur", "ils", "elles", "on", "les", "en", "ci", "là"]

# remplacer les « t » euphoniques par 't_'
nana_euph = nana2.replace("-t-", " t_").replace("-t'", " t_")


# passer à travers chaque mot de la chaîne qui nous intéresse (segmentation préliminaire avec espace)
for mot in nana_euph.split():
   
    # si tiret court dans mot, vérifier ce qui vient à la fin (avec split)
    if "-" in mot :
        apres_tiret = mot.split("-")[-1]
        
        # si ce qui vient avant l'apostrophe est dans la liste de clitiques, ajouter espace après apostrophe
        if apres_tiret in pron_dem :
            mot = mot.replace("-", " ")

    # ajouter le mot (modifié ou non) à la liste finale d'éléments
    nana_liste2.append(mot)
    
# tranformer nana_liste en chaîne avec join()
nana3 = " ".join(nana_liste2)
print(nana3)
```

```python
# extraire les mots

# liste vide 
mots_nana = []

# signes de ponctuation
ponctuation = [".", ",", "!", "?", "...", "…", ":", ";", "«", "»", "(", ")", "—", "–"]

# passer à travers la chaîne segmentée (split) et écarter les tokens qui relèvent de la ponctuation
for token in nana3.split():
    if token not in ponctuation :
        mots_nana.append(token)

print(mots_nana)
```

## 2. STATISTIQUES LEXICALES DE BASE



### 2.1. Occurrences et types

Maintenant que nous avons tokenisé notre texte, il est possible d'extraire des statistiques lexicales de base. Pour y arriver, il faut d'abord distinguer deux notions importantes :
 - **Occurrence** : chaque utilisation individuelle d'un mot (p. ex. *je pense donc je suis* : 5 mots-occurrences)
 - **Type** : chaque forme unique de mot (p. ex. *je pense donc je suis* : 4 mots-types)

Le nombre d'occurrences nous renseigne uniquement sur la longueur du texte. Le nombre de types différents nous donne une certaine idée de la richesse du vocabulaire employé.

#### La casse
- Pour le calcul des occurrences, la casse ne fait pas de différence puisqu'on ne regroupe pas (pour l'instant) les mots en catégories.
- Pour le calcul des types, la casse peut poser problème, car elle pourrait nous mener à traiter « La » et « la » comme deux mots différents. Une solution (imparfaite) consiste à tout mettre en minuscules. Cette approche cause des problèmes pour l'analyse des noms propres, nous y reviendrons...

```python
# nombre de mots (occurrences) avec nouvelle méthode de tokenisation 
print(f"Nombre d'occurrences : {len(mots_nana)}")

# nombre de types différents
nana_types = []

# passer à travers la liste mots_nana, convertir les mots en minuscules
# placer les mots dans nana_types s'ils n'y apparaissent pas déjà
for element in mots_nana :
    element_min = element.lower()
    if element_min not in nana_types :
        nana_types.append(element_min)

print(f"Nombre de types : {len(nana_types)}")

```

## 2.2. Les mots les plus fréquents

Jusqu'ici, nous nous sommes contenté de calculer des totaux pour des textes complets. Nous avons vu que le nombre d'occurrences (dans un texte suffisamment long) est toujours plus élevé que le nombre de types lexicaux, puisque certains mots apparaissent plusieurs fois. Nous allons maintenant essayer de voir plus précisément la fréquence des mots individuels que l'on trouve dans un texte. 

### 2.2.1. Calcul du nombres d'occurrences

Nous pouvons élaborer une petite procédure pour calculer le nombre d'occurrences de chaque mot.
1. Créer un dictionnaire (clés = mots, valeurs = nombre d'occurrences)
2. Chaque fois qu'on rencontre un mot, ajouter 1 à sa valeur

```python
# COMPTER LE NOMBRE D'OCCURRENCES DES MOTS

# créer dictionnaire vide
nb_occ = {}

# passer à travers les mots
# si mot absent du dictionnaire : créer clé avec valeur 1
# si mot présent : ajouter 1 à la valeur 
for mot in mots_nana:
    mot_min = mot.lower()
    if mot_min not in nb_occ :
        nb_occ[mot_min] = 1
    else :
        nb_occ[mot_min] = nb_occ[mot_min] + 1

# imprimer le dictionnaire
print(nb_occ)
        
```

```python
# CLASSER UN DICTIONNAIRE
# Il est assez difficile de consulter les dictionnaire tels quels. 
# Pour nous aider à nous y retrouver, nous pouvons classer les dictionnaire par clé ou par valeur

# par clé
nb_occ_cle = dict(sorted(nb_occ.items()))
print("## Dictionnaire en ordre de clés ##\n", nb_occ_cle)

# par valeur (ne vous souciez pas de la syntaxe exacte)
nb_occ_val = dict(sorted(nb_occ_cle.items(), key=lambda item: item[1], reverse=True))
print("\n\n## Dictionnaire en ordre décroissant de valeurs ##\n", nb_occ_val)
```

### 2.2.2. Nombre absolu et fréquence

Le nombre absolu d'occurrences d'un mot varie évidemment en fonction de la longueur du texte. Cela peut être un problème pour les comparaisons entre textes. Pour cette raison, il peut être préférable d'utiliser une valeur relative : la **fréquence lexicale**, c'est-à-dire le nombre d'occurrences que l'on trouve dans un nombre de mots donné. Une mesure de fréquence couramment employée est le nombre d'occurrences par million de mot.

#### Exemple
- Disons que nous avons un texte de 4567 mots
- Dans ce texte, nous trouvons 2 fois le mot « chapeau »
- On peut faire une règle de trois pour ramener ce ratio sur 1 million
- 1 000 000 * 2 / 4567 = 437,9 occurrences par million de mots

Pour calculer la fréquence d'un mot, il faut donc avoir le nombre total de mots dans le texte, et le nombre d'occurrence du mot qui nous intéresse. Nous pouvons donc modifier le code précédent pour obtenir un calcul de fréquence plutôt qu'un nombre absolu.

```python
# COMPTER LA FRÉQUENCE DES MOTS

# créer dictionnaire vide pour occurrences
nb_occ = {}

# créer dictionnaire vide pour fréquences
freq = {}

# trouver nombre total de mots dans le texte
nb_total = len(mots_nana)

# construire le dictionnaire nb_occ comme vu précédemment
for mot in mots_nana:
    mot_min = mot.lower()
    if mot_min not in nb_occ :
        nb_occ[mot_min] = 1
    else :
        nb_occ[mot_min] = nb_occ[mot_min] + 1

# convertir en fréquences
# pour chaque valeur, rapporter sur 1 000 000, arrondir pour éviter trop écimales
for mot in nb_occ:
    freq_mot = 1000000 * nb_occ[mot] / nb_total
    freq_mot = round(freq_mot, 1)
    freq[mot] = freq_mot



# ordonner le dictionnaire par fréquence (décroissant) et imprimer le dictionnaire
freq_ord = dict(sorted(freq.items(), key=lambda item: item[1], reverse=True))
print(freq_ord)
        
```

## 2.2.3. Les X mots les plus fréquents
- Dans plusieurs cas, on ne s'intéresse pas à l'ensemble du vocabulaire d'un texte, mais seulement aux mots les plus fréquents
- La procédure suivante permet de classer le dictionnaire en ordre décroissant des valeurs, puis de garder seulement les 5, 10, 100, 1000... premiers éléments

```python
# trouver les 5 premiers éléments du dictionnaire ordonné selon valeurs décroissantes (ne vous souciez pas de la syntaxe exacte)
top_5 = dict(sorted(freq.items(), key=lambda item: item[1], reverse=True)[:5])
print(top_5)


# clé des éléments les plus fréquents
cles_freq = []
for mot in top_5:
    cles_freq.append(mot)

print(f"\nLes 5 mots les plus fréquents du texte sont : {cles_freq}")
```
