# Exercice 1 : formatage

Extrait de *Chanson d'automne* de Paul Verlaine (1866).

**Tâches :**
- Reconstituer les sauts de lignes (remplacés par symbole "|" ici)
- Supprimer les espaces en début/fin de vers et uniformiser la casse (majuscule seulement en début de vers, minuscules ailleurs)
- Compter le nombre total de caractères avant et après nettoyage
- Calculer le nombre de voyelles et de consonnes présentes (utiliser les listes vues en classe)

```python
automne = "  Les SANGLOTS longs| Des violons  |De l’AUTOMNE |    Blessent mon COEUR| D’une langueur|Monotone. "



# 1. Compter les caractères avant le nettoyage
caracteres_avant = len(automne)
print("Nombre de caractères avant le nettoyage :", caracteres_avant)

# 2. Découper la strophe en lignes
lignes = automne.split("|")

# preparer une chaîne où placer les résultats
poeme_nettoye = ""

# passer à travers les lignes et 
for ligne in lignes:
    # Supprimer les espaces au début et à la fin
    ligne_propre = ligne.strip()
    
    # Mettre majuscule à première lettre de la ligne (seulement)
    ligne_propre = ligne_propre.capitalize()
    
    # Ajouter la ligne nettoyée au résultat final, avec ajout d'un saut de ligne ("\n")
    poeme_nettoye = poeme_nettoye + ligne_propre + "\n"

# Afficher le poème nettoyé
print("\nPoème nettoyé :")
print(poeme_nettoye)

# Compter le nombre de caractères après le nettoyage
caracteres_apres = len(poeme_nettoye)
print("Nombre de caractères après le nettoyage :", caracteres_apres)

# Compter les voyelles et consonnes

# chaînes contenant toutes les voyelles et consonnes
voyelles = "aeiouyàâäæéèêëïîôùûüÿœAEIOUYÀÂÄÆÉÈÊËÏÎÔÙÛÜŸŒ"
consonnes = "bcdfghjklmnpqrstvwxzçBCDFGHJKLMNPQRSTVWXZÇ"


# Valeurs de base
nombre_voyelles = 0
nombre_consonnes = 0

# Passer à travers le poème nettoyé et, pour chaque caractère, vérifier si consonne/voyelles, mettre compteur à jour
for lettre in poeme_nettoye:
    if caractere in voyelles:
        nombre_voyelles = nombre_voyelles + 1
    if caractere in consonnes:
        nombre_consonnes = nombre_consonnes + 1

# Imprimer résultats finaux
print("\nNombre de voyelles :", nombre_voyelles)
print("Nombre de consonnes :", nombre_consonnes)

















#
```

# Exercice 2 : 

Extrait de *L'Albatros* de Charles Baudelaire (1861).


**Tâches :**
- Pour chaque vers, compter le nombre de mots et la longueur moyenne des mots
- Extraire le dernier mot de chaque vers (pour analyser les rimes)
- Calculez le nombre total de mot se terminant par la lettre 's'

```python
albatros = """Souvent, pour s'amuser, les hommes d'équipage
Prennent des albatros, ces vastes oiseaux des mers
Qui suivent, indolents compagnons de voyage
Le navire glissant sur les gouffres amers"""


# CORRIGÉ



# FACULTATIF
# Si on regarde le détail du code qui suit, on voit que la séparation des mots
# donne des mots du genre "Souvent," qui incluent la virgule
# Pour éviter ce problème, on pourrait se débarasser des virgules 
# (et au besoin autres signes de ponctuation) avec la ligne suivante :
albatros = albatros.replace(",", "")
# On pourrait faire quelque chose du genre avec les apostrophes :
albatros = albatros.replace("'", " ")




# Séparer le poème en lignes
vers_liste = albatros.split("\n")

print("## RÉSULTATS GÉNÉRAUX ###\n")

# Liste où mettre tous les derniers mots
derniers_mots = []

# Numéro du vers (0 par défaut)
no_vers = 0

# Passer à travers les lignes
for vers in vers_liste:

    # obtenir numéro du vers en ajoutant 1 à no_vers
    no_vers = no_vers + 1

    
    print("- Vers", no_vers, ":", vers)
    
    # Séparer la chaîne à chaque espace, et compter les mots
    # Noter : vers.split() est équivalent à vers.split(" ")
    mots = vers.split()  
    nombre_mots = len(mots)
    print("- Nombre de mots :", nombre_mots)
    
    # Calculer la longueur moyenne des mots

    # valeur de base (on y ajoutera progressivement la longueur de chaque mot)
    longueur_totale = 0

    # Passer à travers les mots du vers et ajouter la longueur de chaque mot à la longueur totale
    for mot in mots:
        longueur_totale = longueur_totale + len(mot)

    # Calculer la longueur moyenne des mots et imprimer
    longueur_moyenne = longueur_totale / nombre_mots
    print("- Longueur moyenne des mots :", longueur_moyenne)
    
    # Extraire le dernier mot du vers et l'imprimer, puis l'ajouter à la liste de derniers mots
    dernier_mot = mots[-1]
    print("- Dernier mot (rime) :", dernier_mot)
    derniers_mots.append(dernier_mot)
    print()

# Afficher les rimes
print("### RIMES ###")
print("- Derniers mots de chaque vers :", derniers_mots)
print()

# Calculer le nombre total de mots se terminant par 's'

# valeur de base pour mots se terminant par 's' (0 au départ)
mots_terminant_s = 0

# Passer à travers chaque vers 
for vers in vers_liste:

    # séparer le vers en mots (avec espaces)
    mots = vers.split()

    # Passer à travers chaque mot du vers
    for mot in mots:
        # Vérifier si le dernier caractère est 's' ou 'S'
        derniere_lettre = mot[-1]
        if derniere_lettre == "s" or derniere_lettre == "S":
            mots_terminant_s = mots_terminant_s + 1

print("### STATISTIQUES ###")
print("Nombre total de mots se terminant par 's' :", mots_terminant_s)
















#
```

# Exercice 3 :

Premier paragraphe de *Le Ventre de Paris* d'Émile Zola (1873).

**A) Longueur des phrases :** À l'aide d'une chaîne-f, produire un petit rapport qui précise :
- Le nombre de mots dans la phrase la plus longue
- Le nombre de mots dans la phrase la plus courte
- La longueur moyenne des phrases

**B) Mots fréquents :** Sous forme d'une énumération de mots séparés par le symbole ¤, donner tous les mots apparaissant plus d'une fois dans le texte (p. ex. `poule ¤ oeuf ¤ couver`).

**C) Schéma orthographique :** Identifier tous les mots commençant par une voyelle et se terminant par une consonne. Créer une liste Python de ces mots, sans doublons.

**D) Noms propres potentiels :** Sous forme d'une énumération de mots séparés par le symbole *, donner tous les mots commençant par une majuscule mais n'étant pas en début de phrases (p. ex. `Pauline * Londres * Jacques`).

**Conseils :**
- Penser à se  débarrasser des sauts de lignes, qui ne sont là que pour aider l'affichage.
- Utiliser les points comme séparateurs de phrases


```python
ventre = '''Au milieu du grand silence, et dans le désert de l’avenue, les voitures de maraîchers montaient vers Paris, 
    avec les cahots rythmés de leurs roues, dont les échos battaient les façades des maisons, endormies aux deux bords, 
    derrière les lignes confuses des ormes. Un tombereau de choux et un tombereau de pois, au pont de Neuilly, 
    s’étaient joints aux huit voitures de navets et de carottes qui descendaient de Nanterre ; et les chevaux allaient tout seuls, 
    la tête basse, de leur allure continue et paresseuse, que la montée ralentissait encore. En haut, sur la charge des légumes, 
    allongés à plat ventre, couverts de leur limousine à petites raies noires et grises, les charretiers sommeillaient, 
    les guides aux poignets. Un bec de gaz, au sortir d’une nappe d’ombre, éclairait les clous d’un soulier, la manche bleue d’une blouse, 
    le bout d’une casquette, entrevus dans cette floraison énorme des bouquets rouges des carottes, des bouquets blancs des navets, 
    des verdures débordantes des pois et des choux. Et, sur la route, sur les routes voisines, en avant et en arrière, 
    des ronflements lointains de charrois annonçaient des convois pareils, tout un arrivage traversant les ténèbres et le gros sommeil de deux heures du matin, 
    berçant la ville noire du bruit de cette nourriture qui passait.'''

# CORRIGÉ

# Commencer par retirer les sauts de ligne (toujours suivis par 4 espaces)
texte_propre = ventre.replace("\n    ", "")
print(texte_propre)

# A. Longueur des phrases (phrase + longue, phrase + courte, longueur moyenne)

# Séparer le texte en phrases (avec point comme séparateur)
phrases = texte_propre.split(".")

# Valeurs de base pour phrases + longue, + courte
mots_max = 0
mots_min = 999999
# Valeurs de base pour trouver longueur moyenne
total_mots = 0
nombre_phrases = 0

# Passer à travers chaque phrase
for phrase in phrases:
    # Retirer espaces superflues au début et à la fin
    phrase = phrase.strip()

    # Séparer la phrase en mots (espace comme séparateur) et calculer nombre de mots
    mots = phrase.split()
    nombre_mots = len(mots)

    # écarter éléments de mots qui auraient 0 mots
    if len(mots) > 0 :
    
        # voir si c'est la phrase la plus longue jusqu'ici
        # si oui, remplacer mots_max
        if nombre_mots > mots_max:
            mots_max = nombre_mots
    
        # voir si c'est la phrase la plus courte jusqu'ici
        # si oui, remplacer mots_min
        if nombre_mots < mots_min:
            mots_min = nombre_mots
    
        # ajouter le nombre de mots de la phrase au total jusqu'ici, ajouter 1 à nombre_phrases
        total_mots = total_mots + nombre_mots
        nombre_phrases = nombre_phrases + 1

# calcul de la longueur moyenne
longueur_moyenne = round(total_mots / nombre_phrases, 2)

print("\n### A. RÉSULTATS - LONGUEUR DES PHRASES")
print(f"La phrase la plus longue a {mots_max} mots.")
print(f"La phrase la plus  courte a {mots_min} mots.")
print(f"Les phrases du texte ont en moyenne {longueur_moyenne:} mots.")
print()

# B. Mots fréquents (+ d'une occurrence)

# Séparer le texte complet en mots
tous_les_mots = texte_propre.split()

# listes vides où mettre tous les mots trouvés, puis seulement trouvés plus d'une fois
liste_tout = []
liste_multi = []
compteurs = []

# Passer à travers le texte, mot par mot
for mot in tous_les_mots:
    # Nettoyer le mot (enlever ponctuation)
    mot_propre = mot.replace(".", "").replace(",", "").strip()

    # mettre mot tout en minuscules (éviter doublons du genre Elle/elle)
    mot_minus = mot_propre.lower()

    # Si le mot est déjà dans liste_tout, l'ajouter dans liste_multi, 
    #mais seulement si pas déjà dans liste multi
    if mot_minus in liste_tout and mot_minus not in liste_multi:
        liste_multi.append(mot_minus)
    
    # ajouter mot dans liste_tout
    liste_tout.append(mot_minus)

mots_freq = " ¤ ".join(liste_multi)


print("\n### B. RÉSULTATS - MOTS FRÉQUENTS")

print("Mots apparaissant plus d'une fois :", mots_freq)
print()

# C. Schéma orthographique (mots débutant par voyelle, terminant par consonne)

# chaînes contenant toutes voyelles et consonnes du français
voyelles = "aeiouyàâäæéèêëïîôùûüÿœAEIOUYÀÂÄÆÉÈÊËÏÎÔÙÛÜŸŒ"
consonnes = "bcdfghjklmnpqrstvwxzçBCDFGHJKLMNPQRSTVWXZÇ"

# liste vide pour recevoir les mots correspondant au schéma
mots_schema = []

# passer à travers les mots de liste_tout
for mot in liste_tout:
    # éviter d'inclure éléments vides    
    if len(mot) > 0:
        # identifier première et dernière lettre
        premiere = mot[0]
        derniere = mot[-1]
        
        # Vérifier si première lettre est voyelle et dernière consonne
        if premiere in voyelles and derniere in consonnes:
            #si mot pas déjà dans la liste
            if mot not in mots_schema:
                mots_schema.append(mot)


print("\n### C. RÉSULTATS - SCHÉMA ORTHOGRAPHIQUE")

print("Mots commençant par une voyelle et se terminant par une consonne :")
for mot in mots_schema:
    print("-", mot)

print()

# D. Noms propres potentiels (mots commençant par majuscule, mais pas après point)

# liste de phrases propres à remplir
phrases_propres = []

# mots où mettre les mots à initiale majuscule (sauf début de phrases)
mots_maj = []

# chaîne avec toutes majuscules
majuscules = "AEIOUYÀÂÄÆÉÈÊËÏÎÔÙÛÜŸŒBCDFGHJKLMNPQRSTVWXZÇ"

# Séparer le texte en phrases (avec point comme séparateur)
phrases = texte_propre.split(".")

# passer à travers les phrases, retirer espaces superflues, envoyer dans phrases_propres
for phrase in phrases:
    phrase_propre = phrase.strip()
    phrases_propres.append(phrase_propre)

# Passer à travers chaque phrase propre
for phrase in phrases_propres:
    # Découper la phrase en mots avec espaces
    mots = phrase.split()

    # passer à travers les mots, en écartant le premier (dont la majuscule ne nous intéresse pas)
    for mot in mots[1:]:

        #si la première lettre est majuscule, mettre dans mots_maj
        if mot[0] in majuscules:
            mots_maj.append(mot)

print("\n### D. RÉSULTATS - NOMS PROPRES POTENTIELS")

print("Mots commençant par une majuscule ailleur qu'en début de phrase :")
for mot in mots_maj:
    print("-", mot)



#
```
