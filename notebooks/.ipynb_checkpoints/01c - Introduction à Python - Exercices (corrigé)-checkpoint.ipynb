## Exercice 7 : Boucles for avec listes

Voici une liste de prix littéraires québécois et l'années de création de chacun :



Écrivez une boucle qui :
1. Affiche chaque prix avec son année de création
2. Compte et affiche combien de prix ont été créés après 1960
3. Trouve et affiche le prix le plus ancien

````python
prix_litteraires = {
    "Grand Prix du livre de Montréal" : 1965,
    "Médaille de l'Académie des lettres du Québec" : 1946, 
    "Prix Arthur-Buies" : 1978, 
    "Prix du Cercle du livre de France" : 1949, 
    "Prix du Gouverneur général" : 1959,
    "Prix France-Québec" : 1998,
    "Prix littéraire des collégien·ne·s" : 2004,
    "Prix Octave-Crémazie" : 1980
}


# CORRIGÉ

# Imprimer entête
print("### RÉSULTATS ###")

# 1. Passer à travers chaque item du dictionnaire et imprimer le nom et l'année de création
for prix in prix_litteraires:
    annee = prix_litteraires[prix]
    print("- Le", prix, "a été créé en", annee)

# Il est aussi possible de procéder avec .items(). Retirer les # pour tester
#for prix, annee in prix_litteraires.items():
#    print("Le", prix, "a été créé en", annee)


# 2. Compter et afficher le nombre de prix créés après 1960

# Préparer une variable pour compter le nombre de prix créés après 1960 (valeur initiale : 0)
post1960 = 0

# Passer à travers les prix. Si l'année de création d'un prix > 1960, ajouter 1 à post1960
for prix in prix_litteraires:
    if prix_litteraires[prix] > 1960 :
        post1960 = post1960 + 1

# Imprimer la somme finale de post1960
print("- Nombre de prix créés après 1960 :", post1960)


# 3. Trouver et afficher le prix le plus ancien

# Préparer des variables de base pour enregistrer le prix le plus ancien à chaque étape
# Ici, la valeur de annee_ancien est très élevée, car on veut que le premier prix qui passera dans la boucle, 
# aussi récent soit-il, soit plus bas que la valeur initiale
plus_ancien = ""
annee_ancien = 9999

# Passer à travers les prix. Si l'année de création est plus basse que annee_ancien, mettre à jour plus_ancien et anne_ancien
for prix in prix_litteraires:
    annee = prix_litteraires[prix]
    if annee < annee_ancien : 
        annee_ancien = annee
        plus_ancien = prix

# Imprimer les valeurs finales de plus_ancien et annee_ancien
print("- Prix le plus ancien :", plus_ancien, "(créé en", annee_ancien, ").")

````

## Exercice 8 : Fonctions simples

Réécrivez le code de l'exercice 7 sous forme de trois fonctions : `annee_creation`, `combien_apres_1960` et `prix_plus_ancien`.

````python
# CORRIGÉ 




# Fonction 1. Passer à travers chaque item du dictionnaire et imprimer le nom et l'année de création
def annee_creation(dictionnaire_prix):
    for element in dictionnaire_prix:
        annee = dictionnaire_prix[element]
        print("- Le", prix, "a été créé en", annee)



# Fonction 2. Compter et afficher le nombre de prix créés après 1960
def combien_apres_1960(dictionnaire_prix):
    # Préparer une variable pour compter le nombre de prix créés après 1960 (valeur initiale : 0)
    post1960 = 0
    
    # Passer à travers les prix. Si l'année de création d'un prix > 1960, ajouter 1 à post1960
    for element in dictionnaire_prix:
        if dictionnaire_prix[element] > 1960 :
            post1960 = post1960 + 1
            
    # ici, plutôt que d'imprimer le résultat (chaîne), nous le renvoyons comme résultat de l'application de cette fonction
    return(post1960)



# Fonction 3. Trouver et afficher le prix le plus ancien
def prix_plus_ancien(dictionnaire_prix):
    
    # Préparer des variables de base pour enregistrer le prix le plus ancien à chaque étape
    # Ici, la valeur de annee_ancien est très élevée, car on veut que le premier prix qui passera dans la boucle, 
    # aussi récent soit-il, soit plus bas que la valeur initiale
    plus_ancien = ""
    annee_ancien = 9999

    # Créer un dictionnaire de nous allons utiliser pour renvoyer les résultats (nom du prix + année)
    reponses = {}
    
    # Passer à travers les prix. Si l'année de création est plus basse que annee_ancien, mettre à jour plus_ancien et anne_ancien    
    for element in dictionnaire_prix:
        annee = dictionnaire_prix[element]
        if annee < annee_ancien : 
            annee_ancien = annee
            plus_ancien = prix

    # Placer les réponses finales dans le dictionnaire et le renvoyer
    reponses["prix"] = plus_ancien
    reponses["annee"] = annee_ancien
    return(reponses)

# Début du code

# Appliquer la fonction annee_creation au dictionnaire prix_litteraires
# Cette fonction ne fait qu'imprimer des éléments, elle ne renvoie rien
annee_creation(prix_litteraires)

# Imprimer une barre pour séparer les réponses
print("-----------------------")

# Appliquer la fonction combien_apres_1960 au dictionnaire prix_litteraires
# Cette fonction n'imprime rien d'emblée, elle renvoie une valeur, que nous devons ensuite imprimer explicitement ici
resultat = combien_apres_1960(prix_litteraires)
print("- Nombre de prix créés après 1960 :",resultat)

# Imprimer une barre pour séparer les réponses
print("-----------------------")


# Appliquer la fonction prix_plus_ancien au dictionnaire prix_litteraires
# Cette fonction n'imprime rien d'emblée, elle renvoie un dictionnaire, à partir duquel nous allons extraire les résultats finaux
resultat_ancien = prix_plus_ancien(prix_litteraires)
print("- Prix le plus ancien :", resultat_ancien["prix"], "(créé en", resultat_ancien["annee"], ").")
````

## Exercice 9 : Analyse de données littéraires

Vous disposez des données suivantes sur des romans québécois :

```python
romans_quebecois = [
    {"titre": "Kamouraska", "auteur": "Anne Hébert", "annee": 1970, "pages": 250},
    {"titre": "Les Fous de Bassan", "auteur": "Anne Hébert", "annee": 1982, "pages": 285},
    {"titre": "L'Hiver de force", "auteur": "Réjean Ducharme", "annee": 1973, "pages": 283},
    {"titre": "Volkswagen Blues", "auteur": "Jacques Poulin", "annee": 1984, "pages": 290},
    {"titre": "La Guerre, Yes Sir!", "auteur": "Roch Carrier", "annee": 1968, "pages": 124}
]
```

Écrivez un programme qui :
1. Affiche tous les romans publiés dans les années 1980
2. Calcule et affiche la longueur moyenne des romans
3. Trouve et affiche le roman le plus court et le plus long
4. Compte combien d'auteurs différents sont représentés

````python
# CORRIGÉ

# liste des romans
romans_quebecois = [
    {"titre": "Kamouraska", "auteur": "Anne Hébert", "annee": 1970, "pages": 250},
    {"titre": "Les Fous de Bassan", "auteur": "Anne Hébert", "annee": 1982, "pages": 285},
    {"titre": "L'Hiver de force", "auteur": "Réjean Ducharme", "annee": 1973, "pages": 283},
    {"titre": "Volkswagen Blues", "auteur": "Jacques Poulin", "annee": 1984, "pages": 290},
    {"titre": "La Guerre, Yes Sir!", "auteur": "Roch Carrier", "annee": 1968, "pages": 124}
]


# Imprimer entête
print("### RÉSULTATS ###")


# 1. Afficher tous les romans publiés dans les années 1980

# Liste vide pour romans publiés dans les années 1980
romans_80 = []

# Passer à travers les dictionnaires de la liste
for roman in romans_quebecois:
    # Si la valeur associée à la clé "annee" est entre 1980 et 1989 (inclusivement), ajouter le titre du roman à romans_80
    if roman["annee"] >= 1980 and roman["annee"] <= 1989:
        romans_80.append(roman["titre"])

# Imprimer liste
print("- Roman parus dans les années 1980 :", romans_80)



#2. Calculer et afficher la longueur moyenne des romans 

# valeurs de base 
nb_romans = 0
nb_pages_total = 0

# Passer à travers les dictionnaires de la liste
for roman in romans_quebecois:
    # Pour chaque dictionnaire, ajouter 1 à nb_roman, et ajouter la valeur de "pages" à nb_pages_total
    nb_romans = nb_romans + 1
    nb_pages_total = nb_pages_total + roman["pages"]
    
# Pour obtenir la moyenne : diviser valeur finale de nb_pages_total par nb_romans 
# Imprimer résultat
pages_moy = nb_pages_total / nb_romans
print("- Nombre de pages moyen :", pages_moy)


# 3. Trouver et afficher le roman le plus court et le plus long

# valeurs de base (la valeur de pages_court est très haute car nous cherchons à trouver la valeur la plus basse)
titre_long = ""
pages_long = 0
titre_court = ""
pages_court = 99999

# Passer à travers les dictionnaires de la liste
for roman in romans_quebecois:
    # Pour chaque dictionnaire, vérifier si nombre de pages est plus haut que pages_long
    # si oui, mettre à jour titre_long et pages_long
    if roman["pages"] > pages_long :
        titre_long = roman["titre"]
        pages_long = roman["pages"]
    
# Imprimer résultat
pages_moy = nb_pages_total / nb_romans
print("- Livre le plus long :", titre_long)

# Passer à travers les dictionnaires de la liste
for roman in romans_quebecois:
    # Pour chaque dictionnaire, vérifier si nombre de pages est plus bas que pages_court
    # si oui, mettre à jour titre_long et pages_long
    if roman["pages"] < pages_court :
        titre_court = roman["titre"]
        pages_court = roman["pages"]
    
# Imprimer résultat
pages_moy = nb_pages_total / nb_romans
print("- Livre le plus court :", titre_court)



# 4. Compter combien d'auteurs différents sont représentés

# liste des auteurs (nous n'y ajouterons que les auteurs qui n'y sont pas déjà)
auteurs = []

# Passer à travers les dictionnaires de la liste
for roman in romans_quebecois:
    # Pour chaque dictionnaire, trouver le nom de l'auteur
    # si cet auteur n'est pas dans la liste auteurs, l'ajouter
    if roman["auteur"] not in auteurs :
        auteurs.append(roman["auteur"])
    
# Calculer la taille de la liste et imprimer le résultat
auteurs_diff = len(auteurs)
print("- Nombre d'auteurs différents :", auteurs_diff)
````
