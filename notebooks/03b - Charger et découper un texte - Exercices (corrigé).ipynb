```python
# CORRIGÉ

# importer bibliothèque os (pour traitement des chemins
import os

# liste de chemins
livres_verne = [r"C:\Users\sainhu01\OneDrive - UQAR\Documents\Cours\2025 - Données textuelles\Environnement\Données\CIDRE\Verne\1863_Cinq_Semaines_en_ballon.epub.txt",
                r"C:\Users\sainhu01\OneDrive - UQAR\Documents\Cours\2025 - Données textuelles\Environnement\Données\CIDRE\Verne\1864_Voyage_au_centre_de_la_Terre.epub.txt",
                r"C:\Users\sainhu01\OneDrive - UQAR\Documents\Cours\2025 - Données textuelles\Environnement\Données\CIDRE\Verne\1865_De_la_Terre_a_la_Lune.epub.txt"]


# fonctions

# fonction 1 : nombre de mots differents (dans une chaîne spécifiée)
def nb_mots_diff(chaine) :

    # effacer ponctuations principales, mettre en minuscule
    chaine_nett = chaine.replace(".", "").replace(",", "").replace("!", "").replace("?", "").lower()

    # séparer chaîne où espaces
    mots = chaine_nett.split()

    # liste de mots differents
    mots_diff = []

    # passer à travers mots, mettre dans la liste si pas déjà là
    for mot in mots :
        if mot not in mots_diff:
            mots_diff.append(mot)

    # trouver le nombre d'éléments dans la liste
    nb_mots = len(mots_diff)

    # retourner le résultat
    return(nb_mots)

# fonction 2 : longueur moyenne des mots (dans une chaîne spécifiée)
def long_mots(chaine) : 
    
    # effacer ponctuations principales, mettre en minuscule
    chaine_nett = chaine.replace(".", "").replace(",", "").replace("!", "").replace("?", "").lower()

    # séparer chaîne où espaces
    mots = chaine_nett.split()

    # liste de tous le smots
    liste_mots = []

    # compteur nb total de caractères
    carac = 0

    # passer à travers mots, ajouter nombre de caractères à total, mettre mot dans la liste 
    for mot in mots :
        carac = carac + len(mot)
        liste_mots.append(mot)

    # calculer moyenne + arrondissement pour limiter à un chiffre après la décimale
    moy = carac / len(liste_mots)
    moy_arrondi = round(moy, 1)

    # retourner résultat
    return(moy_arrondi)
    


    
# dictionnaires avec les resultats
res_nb_mots = {}
res_long_mots = {}

# Passer à travers la liste de fichiers et lire chacun 
for fichier in livres_verne :
    # extraire le nom du fichier (sans l'emplacement)
    titre = os.path.basename(fichier)
    
    with open(fichier, 'r', encoding='utf-8') as f:
        # assigner un nom de variable au contenu textuel
        texte = f.read()

    # envoyer résultats des deux fonctions vers les dictionnaires
    res_nb_mots[titre] = nb_mots_diff(texte)
    res_long_mots[titre] = long_mots(texte)



# imprimer rapport 
print("## Nombre de mots différents ##")
for livre in res_nb_mots:
    print(f"- Le fichier «{livre}» compte {res_nb_mots[livre]} mots différents.")

print("\n## Longueur moyenne des mots ##")
for livre in res_long_mots:
    print(f"- Dans le fichier «{livre}», les mots comptent en moyenne {res_long_mots[livre]} caractères.")
```
